```python
import matplotlib .pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    "BCA" :[10, 20, 30, 40, 50],
    "MCA" :[20, 3, 30, 4, 5],
    "Years":[2012, 2013, 2014, 2015, 2016]
})
```

```python
plt.plot(df["Years"], df["BCA"], marker= "o", label= "BCA")
plt.plot(df["Years"], df["MCA"], marker= "^", label= "MCA")
plt.title("Admission Data")
plt.xlabel("Years")
plt.ylabel("No. of admission")
plt.grid()
```
